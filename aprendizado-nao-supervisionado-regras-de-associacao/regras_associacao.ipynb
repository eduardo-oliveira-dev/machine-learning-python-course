# **APRIORI**

```python
from pyECLAT import Example1
dados = Example1().get()
dados
```

```python
dados.shape
```

```python
from pyECLAT import ECLAT
eclat = ECLAT(data=dados, verbose=True)
dados2 = eclat.df_bin
dados2
```

```python
!pip install mlxtend
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# Gerando associacao
associacao = apriori(dados2, min_support=0.05, use_colnames=True)
```

```python
associacao
```

```python
# Colocando a associação em ordem dos mais frequentes
associacao.sort_values(by='support', ascending=False).head(15)
```

```python
# Criando as regras de associação
regras = association_rules(associacao, metric='confidence')
```

```python
resultado = regras.sort_values(by='lift', ascending=False)
```
